# Introduction

In 1897 J. J. Thomson's cathode-ray experiments established that electrons
carry a definite charge-to-mass ratio and exist as free particles.  Only three
years later, barely allowing the dust of the discovery to settle, Paul Drude
proposed a remarkably simple mechanical picture of a metal: an ideal gas of
electrons drifting through a static lattice of heavy positive ions, colliding
occasionally and otherwise free.

From that one picture Drude extracted Ohm's law, the Hall effect, the
Wiedemann-Franz law, and a qualitative account of the optical reflectivity of
metals.  The agreement with experiment was, for its time, astonishing.

The Drude model is also wrong in ways that matter.  Classical statistics
overcounts the electronic specific heat by two orders of magnitude, predicts
the wrong temperature dependence of the resistivity, and assigns to each
electron a mean speed utterly incompatible with what is actually measured in a
Hall bar.  All of these problems vanish once one concedes — as Sommerfeld did
in 1927 — that the conduction electrons obey Fermi-Dirac statistics rather
than Maxwell-Boltzmann.

This notebook develops the classical Drude model from first principles,
derives its predictions for DC and AC response, works out Joule heating in
both regimes, diagnoses the failures, and finally presents the Drude-Sommerfeld
extension and the corrections it brings.  The accompanying computation
notebook implements the same physics numerically.

# The Drude Model

## Assumptions

Drude's starting point is a set of four deliberately crude hypotheses about
the conduction electrons in a metal:

1. *Independent-electron approximation.*  Between collisions electrons do not
   interact with each other or with the lattice ions, except through any
   externally applied electromagnetic field.
2. *Free-electron approximation.*  In the same intervals, electrons move as
   free classical particles obeying Newton's second law.
3. *Collisions are instantaneous events* that randomize an electron's
   velocity.  The probability per unit time of colliding is $1/\tau$, where
   $\tau$ is the **relaxation time** (or *mean free time*).  After a collision
   the electron emerges with a velocity drawn from the thermal distribution
   appropriate to the local temperature, with no memory of its previous
   velocity.
4. *Thermal equilibration at collisions.*  Each collision brings the electron
   into local thermodynamic equilibrium with the lattice.

The number density of mobile electrons is $n$.  For a monovalent metal with
mass density $\rho_m$ and atomic mass $M$,
$n = \rho_m N_A/M \sim 10^{28}\,\text{m}^{-3}$.

## Equation of motion

Averaging assumption 3 over a large ensemble of electrons yields an equation
of motion for the drift (ensemble-averaged) velocity $\vec v$.  In a time
$dt$, the probability that a given electron *has not* collided is
$1 - dt/\tau$, in which case its velocity changes by
$(\vec F/m)\,dt$ under the external force $\vec F$.  The probability that it
*has* collided is $dt/\tau$, in which case its post-collision velocity has
zero mean.  Hence
$$
\langle \vec v(t+dt)\rangle = \left(1-\frac{dt}{\tau}\right)
\left[\vec v(t) + \frac{\vec F}{m}dt\right] + \frac{dt}{\tau}\cdot 0,
$$
which, keeping terms to $O(dt)$, gives the **Drude equation of motion**
$$
m\,\frac{d\vec v}{dt} = \vec F - \frac{m\vec v}{\tau}.
\tag{1}
$$

Equation (1) is a Langevin equation with the stochastic term averaged out:
the linear friction $-m\vec v/\tau$ encodes the *relaxation* induced by
random collisions.  We will restore the stochastic term numerically in the
computation notebook.

# DC Conductivity and Ohm's Law

For a uniform, static electric field $\vec E$, the steady state of (1) is
$d\vec v/dt = 0$, so
$$
\vec v_d = -\frac{e\tau}{m}\vec E,
$$
where $-e$ is the electron charge (we follow the convention $e>0$).  The
resulting current density $\vec J = -ne\vec v_d$ is
$$
\boxed{\;\vec J = \sigma_0 \vec E,\qquad
\sigma_0 = \frac{ne^2\tau}{m}\;}
\tag{2}
$$
which is **Ohm's law** with the Drude DC conductivity $\sigma_0$.  The
resistivity is $\rho = 1/\sigma_0$.

The *mobility* $\mu = e\tau/m$ gives $\vec v_d = -\mu\vec E$; experimentally
one accesses $\mu$ directly via the Hall effect, and $\tau$ then follows.
Typical metallic values at room temperature are $\tau\sim 10^{-14}$ s and
$\mu\sim 10^{-3}\,\mathrm{m^2/(V\,s)}$.

# AC Response

For a harmonic field $\vec E(t) = \vec E_0 e^{-i\omega t}$, substituting the
*ansatz* $\vec v = \vec v_0 e^{-i\omega t}$ into (1) gives
$$
-i\omega m\,\vec v_0 = -e\vec E_0 - \frac{m\vec v_0}{\tau}
\;\Longrightarrow\;
\vec v_0 = -\frac{e\tau/m}{1-i\omega\tau}\vec E_0.
$$
Hence the frequency-dependent **complex conductivity**
$$
\boxed{\;\sigma(\omega) =
\frac{\sigma_0}{1-i\omega\tau}
= \frac{\sigma_0}{1+\omega^2\tau^2}
  + i\,\frac{\sigma_0\,\omega\tau}{1+\omega^2\tau^2}\;}
\tag{3}
$$

Two limits are worth noting.

* *Low frequency,* $\omega\tau\ll 1$: $\sigma\to\sigma_0$ — the DC result.
* *High frequency,* $\omega\tau\gg 1$: $\sigma\to i\sigma_0/(\omega\tau) =
  ine^2/(m\omega)$ — purely imaginary, the response of a collisionless
  electron gas.

Only the *real* part of $\sigma$ contributes to time-averaged Joule heating;
the imaginary part is reactive.

## Dielectric function and the plasma frequency

The relative permittivity of the electron gas follows from
$\epsilon(\omega) = 1 + i\sigma(\omega)/(\epsilon_0\omega)$:
$$
\epsilon(\omega) = 1 - \frac{\omega_p^2}{\omega^2 + i\omega/\tau},
\qquad
\omega_p^2 \equiv \frac{ne^2}{\epsilon_0 m}.
\tag{4}
$$
$\omega_p$ is the **plasma frequency**.  In the collisionless limit
$\tau\to\infty$ the metal is opaque for $\omega<\omega_p$ and transparent for
$\omega>\omega_p$, explaining qualitatively why alkali metals become UV-
transparent above their plasma edge.

# The Hall Effect and Magnetoresistance

In the presence of a magnetic field $\vec B = B\hat z$ the equation of
motion (1) gains a Lorentz term:
$$
m\frac{d\vec v}{dt} = -e(\vec E + \vec v\times\vec B) - \frac{m\vec v}{\tau}.
$$
In steady state with $\vec v = (v_x, v_y, 0)$,
$$
\begin{pmatrix}E_x \\ E_y\end{pmatrix} =
\frac{1}{\sigma_0}\begin{pmatrix}1 & \omega_c\tau \\ -\omega_c\tau & 1\end{pmatrix}
\begin{pmatrix}J_x \\ J_y\end{pmatrix},
$$
where $\omega_c = eB/m$ is the **cyclotron frequency**.  Demanding
$J_y = 0$ (the Hall-bar boundary condition) gives the **Hall coefficient**
$$
R_H \equiv \frac{E_y}{J_x B} = -\frac{1}{ne}.
\tag{5}
$$
A single sign and magnitude, depending *only* on the carrier density.  The
experimental fact that some metals (Be, Zn, Al) have *positive* $R_H$ is
one of Drude's most spectacular failures; its resolution requires band
theory and the notion of hole carriers.

The diagonal component of the resistivity tensor is independent of $B$ in
the Drude model: **the magnetoresistance vanishes.**  This is another
failure in detail, but for the purposes of this project we take equations
(2)–(5) as the classical baseline against which Sommerfeld corrections will
be compared.

# Thermal Conductivity and the Wiedemann-Franz Law

Treating the electrons as an ideal classical gas with mean energy
$\tfrac{3}{2}k_B T$ and mean speed $\bar v$, Drude estimated the thermal
conductivity by kinetic theory:
$$
\kappa = \tfrac{1}{3}\,c_v\,\bar v\,\ell,
$$
where $c_v = \tfrac{3}{2}nk_B$ is the classical electronic specific heat per
unit volume, $\bar v$ the mean thermal speed, and $\ell = \bar v\tau$ the
mean free path.  Substituting $\tfrac{1}{2}m\bar v^2 = \tfrac{3}{2}k_B T$,
$$
\kappa = \frac{n k_B^2 T\tau}{m}\cdot\frac{3}{2}.
$$

Dividing by $\sigma_0 T$ gives the **Wiedemann-Franz ratio**
$$
\boxed{\;\frac{\kappa}{\sigma_0 T} =
L \equiv \frac{3}{2}\left(\frac{k_B}{e}\right)^{\!2}\;}
\tag{6}
$$
— a universal constant, independent of material parameters.  Drude found
numerical agreement to within a factor of two with measurements on copper and
silver, which he famously misreported as *exact* agreement after making two
algebra errors that cancelled.  We will see that Sommerfeld's correction
multiplies the right-hand side by $\pi^2/3$, giving
$L = \pi^2 k_B^2/(3e^2) \approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$
— the value that is actually observed.

# Joule Heating — DC

Consider a conductor of cross-section $A$ and length $L$ carrying a uniform
current $I$ under bias $V = EL$.  The mechanical work done *per unit volume*
per unit time by the field on the electrons is
$$
p = \vec J\cdot\vec E.
\tag{7}
$$
All of this work is ultimately transferred to the lattice through collisions,
because the drift kinetic energy per electron, $\tfrac{1}{2}m v_d^2$, is
stationary in steady state.  Multiplying (7) by the volume $AL$,
$$
P_{\text{DC}} = JE\cdot AL = (JA)(EL) = IV = I^2 R,
\tag{8}
$$
the familiar result.  In a microscopically consistent picture, the work
$e\vec E\cdot\vec v\,dt$ delivered to an electron between collisions is
dissipated at the *next* collision, which resets its velocity to a thermal
value.  This is the sense in which **Joule heating is a consequence of
momentum relaxation**: an idealized ballistic conductor ($\tau\to\infty$)
stores energy rather than dissipating it.

# Joule Heating — AC

For $\vec E(t) = \vec E_0\cos\omega t$ the instantaneous power dissipation
is $p(t) = \vec J(t)\cdot\vec E(t)$.  Writing
$\vec J(t) = \operatorname{Re}\sigma(\omega)\vec E_0\,e^{-i\omega t}$ and
time-averaging over one period,
$$
\langle p\rangle = \tfrac12\,\operatorname{Re}\sigma(\omega)\,|\vec E_0|^2
= \frac{1}{2}\,\frac{\sigma_0}{1+\omega^2\tau^2}\,|\vec E_0|^2.
\tag{9}
$$

Equation (9) contains two important physical statements:

* *Only the real part of* $\sigma$ *dissipates.*  The imaginary part
  corresponds to energy that sloshes between the field and the electron gas
  without heating the lattice.  At $\omega\tau\gg 1$ the dissipation falls as
  $1/\omega^2$: the drift never has time to relax before the field reverses.
* *The resistance lowers at high frequency.*  For RF applications this is a
  welcome effect (the power deposited in the conductor decreases); for
  microwave cavities it competes with skin-effect enhancement.

The *phase lag* between $\vec J$ and $\vec E$ is
$\varphi(\omega) = \arctan(\omega\tau)$, which grows from $0$ at DC to
$\pi/2$ in the collisionless limit.

# Failures of the Classical Drude Model

The Drude picture reproduces the *shape* of many metallic response functions
but fails quantitatively in several places that a careful experimentalist
cannot ignore:

* **Specific heat.**  Equipartition assigns each electron an internal energy
  $\tfrac{3}{2}k_B T$, hence a molar heat capacity of
  $\tfrac{3}{2}R \approx 12.5\,\mathrm{J/(mol\,K)}$ above and beyond the
  lattice contribution.  Measurements on copper give an electronic
  contribution two orders of magnitude smaller and *linear* in $T$.
* **Wiedemann-Franz numerical coefficient.**  The classical
  $L = \tfrac{3}{2}(k_B/e)^2$ is off by a factor of $\pi^2/2$; see (6).
* **Temperature dependence of $\rho$.**  Drude assumes $\tau$ is set by
  collisions with ions; the obvious estimate gives $\tau\propto 1/\sqrt{T}$
  and $\rho\propto\sqrt{T}$.  Measurements give $\rho\propto T$ at high $T$
  and $\rho\propto T^5$ at low $T$.
* **Sign of the Hall coefficient.**  Already discussed.  Several metals show
  $R_H>0$, which is impossible for negatively charged free carriers.
* **Mean electron speed.**  Ohm's law plus (2) gives a drift velocity of
  order millimetres per second, but the *thermal* speed from equipartition
  is $\sim 10^5$ m/s.  Independent measurements of the Fermi velocity give
  $\sim 10^6$ m/s — thirty times larger again, and, critically, *independent
  of temperature*.
* **Thermopower.**  The classical Seebeck coefficient is
  $S = -c_v/(ne) \sim -k_B/e$.  The measured value is two orders of
  magnitude smaller.

Each of these discrepancies will be understood once we introduce Fermi-Dirac
statistics.

# Fermi-Dirac Statistics and the Sommerfeld Correction

Electrons are spin-$\tfrac12$ fermions and therefore obey the Pauli
exclusion principle: no two can occupy the same single-particle quantum
state.  The consequence at finite temperature is that the occupation number
of a state with energy $\varepsilon$ is
$$
f(\varepsilon) = \frac{1}{e^{(\varepsilon-\mu)/k_B T}+1},
\tag{10}
$$
the **Fermi-Dirac distribution**, with chemical potential $\mu$.  At $T=0$,
$f$ is a step function $\Theta(\mu-\varepsilon)$; the value of $\mu$ at zero
temperature is the **Fermi energy** $\varepsilon_F$, and the corresponding
wavevector and velocity are
$$
k_F = (3\pi^2 n)^{1/3},\qquad
v_F = \hbar k_F/m,\qquad
T_F = \varepsilon_F/k_B.
$$
For a typical metal $n\sim 10^{29}\,\text{m}^{-3}$ gives $T_F \sim 10^4$–
$10^5$ K — *far* above room temperature.  Hence in ordinary conditions the
electron gas is **strongly degenerate**: only states within an energy shell
of order $k_B T$ around $\varepsilon_F$ are thermally active.  The
inactive "frozen" core contributes nothing to transport or specific heat.

This single observation repairs most of Drude's quantitative failures:

* Only a fraction $\sim k_B T/\varepsilon_F$ of the electrons contribute to
  $c_v$, giving
  $c_v = \tfrac{\pi^2}{3}\,n k_B\,\frac{k_B T}{\varepsilon_F}$, linear in
  $T$ and small.
* The characteristic electron speed is $v_F$, not $\sqrt{3k_B T/m}$.  The
  mean free path becomes $\ell = v_F\tau$, much larger than Drude thought.
* The Wiedemann-Franz constant acquires a prefactor $\pi^2/3$, yielding the
  experimental Lorenz number.
* The Seebeck coefficient gains an extra factor $k_B T/\varepsilon_F$ and
  becomes small.

# The Drude-Sommerfeld Model

Sommerfeld's prescription is to keep Drude's equation of motion (1) but
replace Maxwell-Boltzmann averaging by Fermi-Dirac averaging.  Transport
quantities are computed by integrals of the form
$$
\langle A \rangle = \int A(\varepsilon)\,g(\varepsilon)\,
\bigl(-\partial_\varepsilon f\bigr)\,d\varepsilon,
\tag{11}
$$
where $g(\varepsilon)$ is the free-electron density of states
$$
g(\varepsilon) = \frac{1}{2\pi^2}\left(\frac{2m}{\hbar^2}\right)^{3/2}
\sqrt{\varepsilon}.
$$
The factor $-\partial_\varepsilon f$ is a peaked function of width $k_B T$
centred at $\mu$.  Expanding $A(\varepsilon)g(\varepsilon)$ around
$\varepsilon_F$ gives the **Sommerfeld expansion**
$$
\int_0^\infty H(\varepsilon)f(\varepsilon)\,d\varepsilon =
\int_0^{\mu}H(\varepsilon)\,d\varepsilon
+ \frac{\pi^2}{6}(k_B T)^2 H'(\mu) + O(T^4).
\tag{12}
$$

Applied to $H = \varepsilon g(\varepsilon)$, equation (12) yields the
linear-in-$T$ electronic specific heat
$$
c_v = \frac{\pi^2}{3}\,n k_B\,\frac{k_B T}{\varepsilon_F}.
\tag{13}
$$
The Wiedemann-Franz ratio becomes
$$
\frac{\kappa}{\sigma_0 T} =
\frac{\pi^2}{3}\left(\frac{k_B}{e}\right)^2 = L_0,
\tag{14}
$$
the **Lorenz number**, $L_0 \approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$.

## Conductivity survives (almost) unchanged

A remarkable consequence of the Pauli principle is that the form of the
Drude conductivity (2)–(3) is *unchanged* in the Sommerfeld model.  Why?
The only states that can be scattered are those within $k_B T$ of the Fermi
surface, but *every* filled state contributes to the current through the
shift $\hbar\vec k \to \hbar\vec k + e\vec E\tau$ of the Fermi sphere in
momentum space.  The net current therefore still comes out as
$\vec J = \sigma_0\vec E$ with $\sigma_0 = ne^2\tau/m$.  Only the
*interpretation* of $\tau$ changes: it is now the relaxation time of the
electrons near the Fermi surface, where the excitations actually live.

## Joule heating in the degenerate gas

Since $\sigma(\omega)$ is unchanged, equations (7)–(9) survive.  The
*mechanism* differs: energy is dissipated through scattering of
quasiparticles near the Fermi surface into unoccupied states also near the
Fermi surface.  The power balance, however, is unchanged, which is why the
distinction is not detectable at the level of a macroscopic resistor.

# Beyond Drude-Sommerfeld

Drude-Sommerfeld shares with its parent a few simplifying lies that must
eventually be paid for.  Three are worth naming:

* The relaxation time $\tau$ is assumed independent of $\vec k$.  In real
  metals $\tau(\vec k)$ varies over the Fermi surface, and different
  scattering channels (phonons, impurities, electron-electron) add via
  Matthiessen's rule.
* The band structure is assumed parabolic.  Replacing $m$ by an effective
  mass $m^*$ absorbs much but not all of the discrepancy; open orbits and
  multiple bands produce qualitatively new effects (magnetic breakdown,
  positive Hall coefficients).
* Electron-electron interactions are ignored.  Landau's Fermi-liquid theory
  shows that at low temperatures these are absorbed into renormalized
  parameters ($m^*$, $v_F^*$, $\tau^*$), leaving the Drude-Sommerfeld
  *form* intact down to remarkably low temperatures.

The hierarchy
$$
\text{Drude} \;\to\; \text{Drude-Sommerfeld} \;\to\;
\text{Bloch-Boltzmann} \;\to\; \text{Fermi liquid}
$$
each step adds one ingredient: quantum statistics, periodic potential,
quasi-particle interactions.  Our computation notebook stops at the second
step and verifies it numerically.

# Summary

| Quantity | Drude | Drude-Sommerfeld | Experiment |
|---|---|---|---|
| $\sigma_0$ | $ne^2\tau/m$ | $ne^2\tau/m$ (same form) | same form |
| $c_v$ | $\tfrac{3}{2}nk_B$ | $\tfrac{\pi^2}{3}\,nk_B\,\frac{k_BT}{\varepsilon_F}$ | $\gamma T$, linear |
| $\kappa/(\sigma_0 T)$ | $\tfrac{3}{2}(k_B/e)^2$ | $\tfrac{\pi^2}{3}(k_B/e)^2$ | $\approx 2.44\times 10^{-8}$ |
| Mean speed | $\sqrt{3k_BT/m}$ | $v_F$ | $\sim 10^6$ m/s |
| $R_H$ | $-1/(ne)$ | $-1/(ne)$ | sign wrong for Be, Zn, Al |

The Drude model is right about the *structure* of metallic response — linear
Ohmic conduction, a Lorentzian conductivity edge at $\omega\tau\sim 1$,
Joule heating proportional to $|E|^2$ — and wrong about any quantity that
depends on *counting* how many electrons actually move, or with what speed.
Sommerfeld's correction is, in a precise sense, the minimal repair:
replace classical counting by quantum counting, and leave the dynamics
alone.  The resulting model is accurate enough that it remains the standard
starting point for metal and plasma electrodynamics well into the
twenty-first century.
