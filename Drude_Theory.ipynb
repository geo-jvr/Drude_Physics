# Introduction

In 1897 J. J. Thomson's cathode-ray experiments established that electrons
carry a definite charge-to-mass ratio and exist as free particles.  Only three
years later, barely allowing the dust of the discovery to settle, Paul Drude
proposed a remarkably simple mechanical picture of a metal: an ideal gas of
electrons drifting through a static lattice of heavy positive ions, colliding
occasionally and otherwise free.

From that one picture Drude extracted Ohm's law, the Hall effect, the
Wiedemann-Franz law, and a qualitative account of the optical reflectivity of
metals.  The agreement with experiment was, for its time, astonishing.

The Drude model is also wrong in ways that matter.  Classical statistics
overcounts the electronic specific heat by two orders of magnitude, predicts
the wrong temperature dependence of the resistivity, and assigns to each
electron a mean speed utterly incompatible with what is actually measured in a
Hall bar.  All of these problems vanish once one concedes — as Sommerfeld did
in 1927 — that the conduction electrons obey Fermi-Dirac statistics rather
than Maxwell-Boltzmann.

This notebook develops the classical Drude model from first principles,
derives its predictions for DC and AC response, works out Joule heating in
both regimes, diagnoses the failures, and finally presents the Drude-Sommerfeld
extension and the corrections it brings.  The accompanying computation
notebook implements the same physics numerically.

## Background Theory: Electron Transport in Metals

To understand electrical and thermal transport in metals, we require a microscopic picture of how electrons move and respond to external fields. The central idea is to treat conduction electrons as mobile charge carriers whose collective motion gives rise to observable macroscopic quantities such as current, conductivity, and heat flow.

In a metal, atoms form a lattice of positive ions, while a subset of electrons are not bound to individual atoms and are free to move throughout the material. These conduction electrons experience two competing effects: acceleration due to external electromagnetic fields, and randomizing collisions with the lattice. The balance between these effects determines the transport properties of the material.

The Drude model provides the simplest framework for describing this behavior by modeling electrons as classical particles undergoing random collisions. The key physical parameter is the **relaxation time** \( \tau \), which represents the average time between collisions. Over timescales shorter than \( \tau \), electrons accelerate under applied forces; over longer timescales, collisions randomize their motion and establish a steady state.

This competition leads naturally to a **drift velocity**, which is the small average velocity superimposed on the otherwise random thermal motion of electrons. Macroscopic electrical current arises from this drift, and transport coefficients such as conductivity can be derived by relating microscopic motion to averaged quantities.

However, the Drude model is fundamentally classical and assumes that electrons obey Maxwell–Boltzmann statistics. While it successfully captures the *form* of many transport laws (such as Ohm’s law), it fails quantitatively because electrons in metals are quantum particles that obey the Pauli exclusion principle.

This motivates the Drude–Sommerfeld model, which retains the dynamical picture of Drude (including relaxation time and equations of motion) but replaces the classical statistical description with **Fermi–Dirac statistics**. In this framework, only electrons near the Fermi energy contribute significantly to transport, leading to corrected predictions for quantities such as heat capacity and thermal conductivity.

Thus, the overall theoretical structure is:
- **Drude model:** Classical dynamics + classical statistics → correct structure, incorrect magnitudes  
- **Sommerfeld correction:** Classical dynamics + quantum statistics → quantitatively accurate transport  

The following sections develop this framework step by step, beginning with the assumptions of the Drude model and progressing toward its quantum-mechanical refinement.

# The Drude Model

## Assumptions

Drude's starting point is a set of four deliberately crude hypotheses about
the conduction electrons in a metal:

1. *Independent-electron approximation.*  Between collisions electrons do not
   interact with each other or with the lattice ions, except through any
   externally applied electromagnetic field.
2. *Free-electron approximation.*  In the same intervals, electrons move as
   free classical particles obeying Newton's second law.
3. *Collisions are instantaneous events* that randomize an electron's
   velocity.  The probability per unit time of colliding is $1/\tau$, where
   $\tau$ is the **relaxation time** (or *mean free time*).  After a collision
   the electron emerges with a velocity drawn from the thermal distribution
   appropriate to the local temperature, with no memory of its previous
   velocity.
4. *Thermal equilibration at collisions.*  Each collision brings the electron
   into local thermodynamic equilibrium with the lattice.

The number density of mobile electrons is $n$.  For a monovalent metal with
mass density $\rho_m$ and atomic mass $M$,
$n = \rho_m N_A/M \sim 10^{28}\,\text{m}^{-3}$.

## Equation of motion

Averaging assumption 3 over a large ensemble of electrons yields an equation
of motion for the drift (ensemble-averaged) velocity $\vec v$.  In a time
$dt$, the probability that a given electron *has not* collided is
$1 - dt/\tau$, in which case its velocity changes by
$(\vec F/m)\,dt$ under the external force $\vec F$.  The probability that it
*has* collided is $dt/\tau$, in which case its post-collision velocity has
zero mean.  Hence
$$
\langle \vec v(t+dt)\rangle = \left(1-\frac{dt}{\tau}\right)
\left[\vec v(t) + \frac{\vec F}{m}dt\right] + \frac{dt}{\tau}\cdot 0,
$$
which, keeping terms to $O(dt)$, gives the **Drude equation of motion**
$$
m\,\frac{d\vec v}{dt} = \vec F - \frac{m\vec v}{\tau}.
\tag{1}
$$

Equation (1) is a Langevin equation with the stochastic term averaged out:
the linear friction $-m\vec v/\tau$ encodes the *relaxation* induced by
random collisions.  We will restore the stochastic term numerically in the
computation notebook.

# DC Conductivity and Ohm's Law

Using the **Drude equation of motion** derived above, we now determine how electrons respond to a uniform, static **electric field** $\mathbf{E}$. Physically, the field attempts to continuously accelerate electrons, but this acceleration is opposed by **random collisions with the lattice**, which relax their momentum over a characteristic time $\tau$. In steady state, these two effects balance, so the average acceleration vanishes, $\frac{d\mathbf{v}}{dt} = 0$. Substituting this condition into the equation of motion gives

$$
\mathbf{v}_d = -\frac{e\tau}{m}\mathbf{E},
$$

where $-e$ is the **electron charge** (with $e > 0$). This expression shows that electrons do not accelerate indefinitely; instead, they acquire a constant **drift velocity** determined by how long they can accelerate before being scattered. The corresponding **current density** $\mathbf{J} = -ne\mathbf{v}_d$, where $n$ is the **number density of conduction electrons**, is therefore

$$
\boxed{
\mathbf{J} = \sigma_0 \mathbf{E}, \quad \sigma_0 = \frac{ne^2\tau}{m}
}
$$

which is **Ohm’s law** with the Drude **DC conductivity** $\sigma_0$. This result makes explicit that electrical conduction is governed by the interplay between **external driving by the electric field** and **scattering processes encoded in $\tau$**. The **resistivity** is $\rho = \frac{1}{\sigma_0}$, so stronger scattering (smaller $\tau$) reduces conductivity. It is also useful to define the **mobility** $\mu = \frac{e\tau}{m}$, giving $\mathbf{v}_d = -\mu \mathbf{E}$; experimentally, $\mu$ is measured via the **Hall effect**, allowing $\tau$ to be inferred.

Thus, Ohm’s law is not assumed, but emerges directly from the Drude dynamics in the steady-state limit. This result represents the **low-frequency limit** of the electron response, which we now generalize to **time-dependent (AC) fields**.

# AC Response

For a harmonic field $\vec E(t) = \vec E_0 e^{-i\omega t}$, substituting the
*ansatz* $\vec v = \vec v_0 e^{-i\omega t}$ into (1) gives
$$
-i\omega m\,\vec v_0 = -e\vec E_0 - \frac{m\vec v_0}{\tau}
\;\Longrightarrow\;
\vec v_0 = -\frac{e\tau/m}{1-i\omega\tau}\vec E_0.
$$
Hence the frequency-dependent **complex conductivity**
$$
\boxed{\;\sigma(\omega) =
\frac{\sigma_0}{1-i\omega\tau}
= \frac{\sigma_0}{1+\omega^2\tau^2}
  + i\,\frac{\sigma_0\,\omega\tau}{1+\omega^2\tau^2}\;}
\tag{3}
$$

Two limits are worth noting.

* *Low frequency,* $\omega\tau\ll 1$: $\sigma\to\sigma_0$ — the DC result.
* *High frequency,* $\omega\tau\gg 1$: $\sigma\to i\sigma_0/(\omega\tau) =
  ine^2/(m\omega)$ — purely imaginary, the response of a collisionless
  electron gas.

Only the *real* part of $\sigma$ contributes to time-averaged Joule heating;
the imaginary part is reactive.

## Dielectric function and the plasma frequency

The relative permittivity of the electron gas follows from
$\epsilon(\omega) = 1 + i\sigma(\omega)/(\epsilon_0\omega)$:
$$
\epsilon(\omega) = 1 - \frac{\omega_p^2}{\omega^2 + i\omega/\tau},
\qquad
\omega_p^2 \equiv \frac{ne^2}{\epsilon_0 m}.
\tag{4}
$$
$\omega_p$ is the **plasma frequency**.  In the collisionless limit
$\tau\to\infty$ the metal is opaque for $\omega<\omega_p$ and transparent for
$\omega>\omega_p$, explaining qualitatively why alkali metals become UV-
transparent above their plasma edge.

## AC Response

Building on the **DC result**, which describes the steady-state (low-frequency) limit, we now consider the response of electrons to a **time-dependent electric field**. For a harmonic field $\mathbf{E}(t) = \mathbf{E}_0 e^{-i\omega t}$, the electron velocity is expected to follow the same time dependence, so we take the ansatz $\mathbf{v}(t) = \mathbf{v}_0 e^{-i\omega t}$. Substituting into the **Drude equation of motion** gives

$$
-i\omega m \mathbf{v}_0 = -e\mathbf{E}_0 - \frac{m}{\tau}\mathbf{v}_0
\quad \Rightarrow \quad
\mathbf{v}_0 = \frac{e\tau/m}{1 - i\omega\tau}\mathbf{E}_0.
$$

This shows that, unlike the DC case, the response now depends on frequency: electrons cannot fully respond to rapidly oscillating fields because their motion is interrupted by **collisions occurring on the timescale $\tau$**. The resulting **current density** leads to the frequency-dependent **complex conductivity**

$$
\boxed{
\sigma(\omega) = \frac{\sigma_0}{1 - i\omega\tau}
= \frac{\sigma_0}{1 + \omega^2\tau^2} + i \frac{\sigma_0 \omega\tau}{1 + \omega^2\tau^2}
}
$$

where the real and imaginary parts have distinct physical meanings:
- the **real part** represents **dissipative transport** (energy loss → Joule heating)  
- the **imaginary part** represents **reactive response** (energy stored and returned to the field)

Two important limits follow naturally:
- **Low frequency** $(\omega\tau \ll 1)$: $\sigma(\omega) \to \sigma_0$, recovering the **DC result**  
- **High frequency** $(\omega\tau \gg 1)$: $\sigma(\omega) \to \frac{i ne^2}{m\omega}$, corresponding to a nearly **collisionless electron gas**

Thus, the AC response generalizes Ohm’s law by showing that conduction is not instantaneous but governed by a finite **relaxation time $\tau$**. Importantly, only the **real part of $\sigma$** contributes to time-averaged **energy dissipation**, linking this section directly to **Joule heating** and, more broadly, to thermal transport.

This frequency-dependent conductivity also determines the material’s electromagnetic response, which we now describe in terms of the **dielectric function** and the **plasma frequency**.

## The Hall Effect and Magnetoresistance

Extending the **AC response framework** to include magnetic fields, we now examine how electron transport is modified in the presence of a static **magnetic field** $\mathbf{B} = B\hat{z}$. In this case, the **Drude equation of motion** acquires an additional **Lorentz force** term,

$$
m\frac{d\mathbf{v}}{dt} = -e(\mathbf{E} + \mathbf{v} \times \mathbf{B}) - \frac{m}{\tau}\mathbf{v}.
$$

While the electric field drives electrons along the conductor, the magnetic field deflects them **perpendicularly** to their motion. This transverse deflection causes charge to accumulate on the sides of the material, generating a **transverse electric field** that balances the Lorentz force in steady state.

For steady transport with $\mathbf{v} = (v_x, v_y, 0)$, solving the coupled equations gives

$$
\begin{pmatrix}
E_x \\
E_y
\end{pmatrix}
=
\frac{1}{\sigma_0}
\begin{pmatrix}
1 & \omega_c \tau \\
-\omega_c \tau & 1
\end{pmatrix}
\begin{pmatrix}
J_x \\
J_y
\end{pmatrix},
$$

where $\omega_c = \frac{eB}{m}$ is the **cyclotron frequency**, which sets the scale of electron deflection in a magnetic field. Imposing the Hall-bar condition $J_y = 0$ isolates the transverse response and gives the **Hall coefficient**

$$
\boxed{
R_H = \frac{E_y}{J_x B} = -\frac{1}{ne}
}
$$

which depends only on the **carrier density**. This shows that the Hall effect provides a direct measurement of both the **sign and density of charge carriers**, making it a fundamental experimental probe of electronic transport.

However, the prediction that $R_H$ is always negative highlights a limitation of the Drude model. Experimentally, some materials exhibit **positive Hall coefficients**, indicating the presence of **hole-like carriers**, which requires a quantum description beyond classical Drude theory.

Finally, the diagonal component of the resistivity tensor is independent of $B$ in the Drude model, so the **magnetoresistance vanishes**. This reflects the model’s simplicity and motivates the need for more refined treatments introduced in the subsequent sections.

# Thermal Conductivity and the Wiedemann-Franz Law

Treating the electrons as an ideal classical gas with mean energy
$\tfrac{3}{2}k_B T$ and mean speed $\bar v$, Drude estimated the thermal
conductivity by kinetic theory:
$$
\kappa = \tfrac{1}{3}\,c_v\,\bar v\,\ell,
$$
where $c_v = \tfrac{3}{2}nk_B$ is the classical electronic specific heat per
unit volume, $\bar v$ the mean thermal speed, and $\ell = \bar v\tau$ the
mean free path.  Substituting $\tfrac{1}{2}m\bar v^2 = \tfrac{3}{2}k_B T$,
$$
\kappa = \frac{n k_B^2 T\tau}{m}\cdot\frac{3}{2}.
$$

Dividing by $\sigma_0 T$ gives the **Wiedemann-Franz ratio**
$$
\boxed{\;\frac{\kappa}{\sigma_0 T} =
L \equiv \frac{3}{2}\left(\frac{k_B}{e}\right)^{\!2}\;}
\tag{6}
$$
— a universal constant, independent of material parameters.  Drude found
numerical agreement to within a factor of two with measurements on copper and
silver, which he famously misreported as *exact* agreement after making two
algebra errors that cancelled.  We will see that Sommerfeld's correction
multiplies the right-hand side by $\pi^2/3$, giving
$L = \pi^2 k_B^2/(3e^2) \approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$
— the value that is actually observed.

# Joule Heating — DC

Consider a conductor of cross-section $A$ and length $L$ carrying a uniform
current $I$ under bias $V = EL$.  The mechanical work done *per unit volume*
per unit time by the field on the electrons is
$$
p = \vec J\cdot\vec E.
\tag{7}
$$
All of this work is ultimately transferred to the lattice through collisions,
because the drift kinetic energy per electron, $\tfrac{1}{2}m v_d^2$, is
stationary in steady state.  Multiplying (7) by the volume $AL$,
$$
P_{\text{DC}} = JE\cdot AL = (JA)(EL) = IV = I^2 R,
\tag{8}
$$
the familiar result.  In a microscopically consistent picture, the work
$e\vec E\cdot\vec v\,dt$ delivered to an electron between collisions is
dissipated at the *next* collision, which resets its velocity to a thermal
value.  This is the sense in which **Joule heating is a consequence of
momentum relaxation**: an idealized ballistic conductor ($\tau\to\infty$)
stores energy rather than dissipating it.

# Joule Heating — AC

For $\vec E(t) = \vec E_0\cos\omega t$ the instantaneous power dissipation
is $p(t) = \vec J(t)\cdot\vec E(t)$.  Writing
$\vec J(t) = \operatorname{Re}\sigma(\omega)\vec E_0\,e^{-i\omega t}$ and
time-averaging over one period,
$$
\langle p\rangle = \tfrac12\,\operatorname{Re}\sigma(\omega)\,|\vec E_0|^2
= \frac{1}{2}\,\frac{\sigma_0}{1+\omega^2\tau^2}\,|\vec E_0|^2.
\tag{9}
$$

Equation (9) contains two important physical statements:

* *Only the real part of* $\sigma$ *dissipates.*  The imaginary part
  corresponds to energy that sloshes between the field and the electron gas
  without heating the lattice.  At $\omega\tau\gg 1$ the dissipation falls as
  $1/\omega^2$: the drift never has time to relax before the field reverses.
* *The resistance lowers at high frequency.*  For RF applications this is a
  welcome effect (the power deposited in the conductor decreases); for
  microwave cavities it competes with skin-effect enhancement.

The *phase lag* between $\vec J$ and $\vec E$ is
$\varphi(\omega) = \arctan(\omega\tau)$, which grows from $0$ at DC to
$\pi/2$ in the collisionless limit.

# Failures of the Classical Drude Model

The Drude picture reproduces the *shape* of many metallic response functions
but fails quantitatively in several places that a careful experimentalist
cannot ignore:

* **Specific heat.**  Equipartition assigns each electron an internal energy
  $\tfrac{3}{2}k_B T$, giving a molar electronic heat capacity
  $\tfrac{3}{2}R \approx 12.5\,\mathrm{J/(mol\,K)}$ — comparable to the
  lattice and easily measurable.  Experiment on copper gives
  $c_v^{(e)} = \gamma T$ with $\gamma \approx 0.695\,\mathrm{mJ/(mol\,K^2)}$:
  two orders of magnitude smaller and *linear* in $T$, not constant.

* **Wiedemann–Franz numerical coefficient.**  The classical Lorenz number
  $L_\text{Drude} = \tfrac{3}{2}(k_B/e)^2 \approx 1.11\times10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$
  is off from the measured value
  $L_\text{exp} \approx 2.44\times10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$
  by a factor of $\pi^2/4.5 \approx 2.2$.

* **Temperature dependence of $\rho$.**  The geometric ion cross-section
  argument gives $\tau\propto 1/\sqrt{T}$ and $\rho\propto\sqrt{T}$.
  Measurements give $\rho\propto T$ at high $T$ and $\rho\propto T^5$
  at low $T$ (Bloch–Grüneisen).

* **Sign of the Hall coefficient.**  Several metals (Be, Zn, Al) show
  $R_H > 0$, impossible for negatively charged free carriers.
  Band theory resolves this through positively charged hole carriers.

* **Mean electron speed.**  Equipartition gives
  $\bar v \sim 10^5\,\mathrm{m/s}$; Hall-bar measurements infer effective
  speeds $\sim 10^6\,\mathrm{m/s}$, independent of temperature —
  a clear sign the relevant speed is a quantum zero-point quantity, not
  a thermal average.

* **Thermopower.**  The classical Seebeck coefficient
  $S \sim -k_B/e \approx -86\,\mu\mathrm{V/K}$ is two orders of magnitude
  larger than the measured $1$–$10\,\mu\mathrm{V/K}$.

Each of these discrepancies will be understood once we replace
Maxwell–Boltzmann with Fermi-Dirac statistics.

# Fermi-Dirac Statistics and the Sommerfeld Correction

Electrons are spin-$\tfrac{1}{2}$ fermions and obey the Pauli exclusion
principle: no two can occupy the same quantum state $(\vec k, s)$.
At finite temperature, the occupation of a state with energy $\varepsilon$ is

$$
f(\varepsilon) = \frac{1}{e^{(\varepsilon-\mu)/k_B T}+1},
\tag{10}
$$

the **Fermi-Dirac distribution**.  At $T=0$, $f$ is a step function cutting
off at the **Fermi energy** $\varepsilon_F$, with

$$
k_F = (3\pi^2 n)^{1/3},\qquad
v_F = \hbar k_F/m,\qquad
T_F = \varepsilon_F/k_B.
\tag{11}
$$

For copper ($n = 8.5\times10^{28}\,\mathrm{m^{-3}}$), $T_F \approx 81{,}000\,\mathrm{K}$,
so at room temperature $T/T_F \approx 0.004 \ll 1$: the gas is **strongly
degenerate**.  Only electrons within a shell of width $\sim k_B T$ around
$\varepsilon_F$ are thermally active — roughly $1\%$ of the total.

## The Sommerfeld expansion

For any smooth $H(\varepsilon)$, integration by parts gives

$$
\int_0^\infty H(\varepsilon)\,f(\varepsilon)\,d\varepsilon
= \int_0^{\mu} H(\varepsilon)\,d\varepsilon
+ \frac{\pi^2}{6}(k_B T)^2\,H'(\mu)
+ \mathcal{O}(T^4).
\tag{12}
$$

The correction is suppressed by $(k_BT/\varepsilon_F)^2 \sim 10^{-5}$ for
copper at room temperature.

## Electronic specific heat

Applying (12) to $H = \varepsilon\,g(\varepsilon)$ gives the mean energy
density $u \approx u_0 + \tfrac{\pi^2}{6}(k_BT)^2 g(\varepsilon_F)$, where
$u_0 = \tfrac{3}{5}n\varepsilon_F$ is the zero-temperature ground-state
energy.  Differentiating:

$$
\boxed{c_v = \frac{\pi^2}{3}\,nk_B\,\frac{k_BT}{\varepsilon_F}}
= \frac{\pi^2}{2}\,k_B\,\theta,
\qquad \theta \equiv \frac{k_BT}{\varepsilon_F} \ll 1.
\tag{13}
$$

This is linear in $T$ and smaller than the Drude value by $\tfrac{\pi^2}{3}\theta
\approx 0.033$ — matching the measured $\gamma T$ behaviour.

## Corrected Lorenz number

Replacing $\bar v \to v_F$ and $c_v \to \tfrac{\pi^2}{2}k_B\theta$ in the
thermal conductivity estimate gives

$$
\boxed{\frac{\kappa}{\sigma_0 T} = L_0
\equiv \frac{\pi^2}{3}\!\left(\frac{k_B}{e}\right)^{\!2}
\approx 2.44\times10^{-8}\,\mathrm{W\,\Omega\,K^{-2}},}
\tag{14}
$$

the experimental Lorenz number, in agreement with Cu, Ag, and Au to within
a few percent.

## Why $\sigma(\omega)$ is unchanged

An external field shifts the *entire* Fermi sphere by
$\delta\vec k = -e\vec E\tau/\hbar$.  The resulting current
$\vec J = -(e/V)\sum_{\vec k}\hbar\vec k\,f(\varepsilon_{\vec k})$
depends only on $n$ and $\tau$, not on the shape of $f$, so

$$
\sigma(\omega) = \frac{\sigma_0}{1-i\omega\tau}, \qquad
\sigma_0 = \frac{ne^2\tau}{m}
\tag{15}
$$

is unchanged.  The Drude formula was accidentally right: correct form,
wrong physical reason.  The remaining failures (thermopower, $\rho(T)$,
Hall sign) require going beyond free electrons entirely — band structure
and phonon physics — which is the subject of the next section.

# The Drude-Sommerfeld Model

Sommerfeld's prescription is to keep Drude's equation of motion (1) but
replace Maxwell-Boltzmann averaging by Fermi-Dirac averaging.  Transport
quantities are computed by integrals of the form
$$
\langle A \rangle = \int A(\varepsilon)\,g(\varepsilon)\,
\bigl(-\partial_\varepsilon f\bigr)\,d\varepsilon,
\tag{11}
$$
where $g(\varepsilon)$ is the free-electron density of states
$$
g(\varepsilon) = \frac{1}{2\pi^2}\left(\frac{2m}{\hbar^2}\right)^{3/2}
\sqrt{\varepsilon}.
$$
The factor $-\partial_\varepsilon f$ is a peaked function of width $k_B T$
centred at $\mu$.  Expanding $A(\varepsilon)g(\varepsilon)$ around
$\varepsilon_F$ gives the **Sommerfeld expansion**
$$
\int_0^\infty H(\varepsilon)f(\varepsilon)\,d\varepsilon =
\int_0^{\mu}H(\varepsilon)\,d\varepsilon
+ \frac{\pi^2}{6}(k_B T)^2 H'(\mu) + O(T^4).
\tag{12}
$$

Applied to $H = \varepsilon g(\varepsilon)$, equation (12) yields the
linear-in-$T$ electronic specific heat
$$
c_v = \frac{\pi^2}{3}\,n k_B\,\frac{k_B T}{\varepsilon_F}.
\tag{13}
$$
The Wiedemann-Franz ratio becomes
$$
\frac{\kappa}{\sigma_0 T} =
\frac{\pi^2}{3}\left(\frac{k_B}{e}\right)^2 = L_0,
\tag{14}
$$
the **Lorenz number**, $L_0 \approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$.

## Conductivity survives (almost) unchanged

A remarkable consequence of the Pauli principle is that the form of the
Drude conductivity (2)–(3) is *unchanged* in the Sommerfeld model.  Why?
The only states that can be scattered are those within $k_B T$ of the Fermi
surface, but *every* filled state contributes to the current through the
shift $\hbar\vec k \to \hbar\vec k + e\vec E\tau$ of the Fermi sphere in
momentum space.  The net current therefore still comes out as
$\vec J = \sigma_0\vec E$ with $\sigma_0 = ne^2\tau/m$.  Only the
*interpretation* of $\tau$ changes: it is now the relaxation time of the
electrons near the Fermi surface, where the excitations actually live.

## Joule heating in the degenerate gas

Since $\sigma(\omega)$ is unchanged, equations (7)–(9) survive.  The
*mechanism* differs: energy is dissipated through scattering of
quasiparticles near the Fermi surface into unoccupied states also near the
Fermi surface.  The power balance, however, is unchanged, which is why the
distinction is not detectable at the level of a macroscopic resistor.

# Beyond Drude-Sommerfeld

Drude-Sommerfeld shares with its parent a few simplifying lies that must
eventually be paid for.  Three are worth naming:

* The relaxation time $\tau$ is assumed independent of $\vec k$.  In real
  metals $\tau(\vec k)$ varies over the Fermi surface, and different
  scattering channels (phonons, impurities, electron-electron) add via
  Matthiessen's rule.
* The band structure is assumed parabolic.  Replacing $m$ by an effective
  mass $m^*$ absorbs much but not all of the discrepancy; open orbits and
  multiple bands produce qualitatively new effects (magnetic breakdown,
  positive Hall coefficients).
* Electron-electron interactions are ignored.  Landau's Fermi-liquid theory
  shows that at low temperatures these are absorbed into renormalized
  parameters ($m^*$, $v_F^*$, $\tau^*$), leaving the Drude-Sommerfeld
  *form* intact down to remarkably low temperatures.

The hierarchy
$$
\text{Drude} \;\to\; \text{Drude-Sommerfeld} \;\to\;
\text{Bloch-Boltzmann} \;\to\; \text{Fermi liquid}
$$
each step adds one ingredient: quantum statistics, periodic potential,
quasi-particle interactions.  Our computation notebook stops at the second
step and verifies it numerically.

# Summary

| Quantity | Drude | Drude-Sommerfeld | Experiment |
|---|---|---|---|
| $\sigma_0$ | $ne^2\tau/m$ | $ne^2\tau/m$ (same form) | same form |
| $c_v$ | $\tfrac{3}{2}nk_B$ | $\tfrac{\pi^2}{3}\,nk_B\,\frac{k_BT}{\varepsilon_F}$ | $\gamma T$, linear |
| $\kappa/(\sigma_0 T)$ | $\tfrac{3}{2}(k_B/e)^2$ | $\tfrac{\pi^2}{3}(k_B/e)^2$ | $\approx 2.44\times 10^{-8}$ |
| Mean speed | $\sqrt{3k_BT/m}$ | $v_F$ | $\sim 10^6$ m/s |
| $R_H$ | $-1/(ne)$ | $-1/(ne)$ | sign wrong for Be, Zn, Al |

The Drude model is right about the *structure* of metallic response — linear
Ohmic conduction, a Lorentzian conductivity edge at $\omega\tau\sim 1$,
Joule heating proportional to $|E|^2$ — and wrong about any quantity that
depends on *counting* how many electrons actually move, or with what speed.
Sommerfeld's correction is, in a precise sense, the minimal repair:
replace classical counting by quantum counting, and leave the dynamics
alone.  The resulting model is accurate enough that it remains the standard
starting point for metal and plasma electrodynamics well into the
twenty-first century.
