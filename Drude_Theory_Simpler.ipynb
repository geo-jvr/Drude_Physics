# Introduction

In 1897 J. J. Thomson's cathode-ray experiments showed that electrons carry
a definite charge-to-mass ratio and exist as free particles. Three years
later, Paul Drude proposed a remarkably simple mechanical picture of a
metal: an ideal *gas* of electrons drifting through a static lattice of
heavy positive ions, colliding occasionally and otherwise free.

From that one picture Drude extracted Ohm's law, the Hall effect, the
Wiedemann-Franz law, and a qualitative account of the optical reflectivity
of metals. The agreement with experiment was, for its time, astonishing.

The Drude model is also wrong in ways that matter. Classical statistics
overcounts the electronic specific heat by two orders of magnitude,
predicts the wrong temperature dependence of the resistivity, and assigns
to each electron a mean speed an order of magnitude smaller than what is
actually measured in a Hall bar. All of these problems vanish once one
concedes --- as Sommerfeld did in 1927 --- that the conduction electrons
obey **Fermi-Dirac statistics** rather than Maxwell-Boltzmann.

This notebook develops the classical Drude model from first principles,
derives its predictions for DC and AC response, works out Joule heating
in both regimes, diagnoses the failures, and finally presents the
**Drude-Sommerfeld** extension and the corrections it brings. The
accompanying **computation notebook** (`Drude_Computation.ipynb`)
implements the same physics numerically and verifies every closed-form
result that follows.

## Background Theory: Electron Transport in Metals

Before any equations, here is the physical picture in plain English.

A metal is a lattice of positive ions sitting in a sea of conduction
electrons. The ions are heavy and stay put; the electrons are light, mobile,
and roughly $10^{28}$ per cubic metre. When you apply an electric field
across the metal, two things happen at once:

1. The field pushes the electrons --- they want to accelerate forever.
2. The electrons keep bumping into things (lattice vibrations, impurities,
   each other) and forgetting which way they were going.

The Drude model rolls (2) into a single number, the **relaxation time**
$\tau$: the average time between collisions. On timescales shorter than
$\tau$ an electron behaves like a free Newtonian particle. On timescales
longer than $\tau$, collisions wash out its memory.

The compromise between "field accelerates" and "collisions reset" produces
a small steady **drift velocity** on top of the otherwise random thermal
motion. That drift is the macroscopic current. Every transport coefficient
in this notebook --- conductivity, Hall coefficient, thermal conductivity,
plasma frequency --- comes from balancing these two effects.

The Drude model gets the *form* of every transport law right (Ohm's law,
the Hall effect, etc.), but it gets several *magnitudes* wrong because real
electrons are quantum particles obeying the Pauli exclusion principle. The
**Drude-Sommerfeld** model fixes this by keeping the same dynamics but
swapping Maxwell-Boltzmann statistics for **Fermi-Dirac statistics**. With
that one change, only electrons within $k_B T$ of the Fermi energy actually
move, and the wrong magnitudes correct themselves.

In summary:

- **Drude:** classical dynamics + classical statistics $\rightarrow$ correct
  *form*, wrong *magnitudes*.
- **Drude-Sommerfeld:** classical dynamics + quantum statistics
  $\rightarrow$ quantitatively accurate transport.

The sections that follow build this picture step by step.

# The Drude Model

## Assumptions

Drude's starting point is four deliberately crude hypotheses about the
conduction electrons in a metal:

1. **Independent-electron approximation.** Between collisions, electrons do
   not interact with each other or with the lattice ions, except through any
   externally applied electromagnetic field.
2. **Free-electron approximation.** In those same intervals, electrons move
   as free classical particles obeying Newton's second law.
3. **Instantaneous, randomizing collisions.** The probability per unit time
   of colliding is $1/\tau$, where $\tau$ is the **relaxation time** (or
   *mean free time*). After a collision the electron emerges with a velocity
   drawn from the local thermal distribution, with *no memory* of its
   previous velocity.
4. **Thermal equilibration at collisions.** Each collision brings the
   electron back into local thermodynamic equilibrium with the lattice.

The number density of mobile electrons is $n$. For a monovalent metal of
mass density $\rho_m$ and atomic mass $M$,
$n = \rho_m N_A / M \sim 10^{28}\,\text{m}^{-3}$.

## Equation of motion

Average assumption 3 over a large ensemble of electrons. In a time $dt$ the
probability that a given electron has *not* collided is $1 - dt/\tau$, in
which case its velocity changes by $(\vec F/m)\,dt$ under the external
force $\vec F$. The probability that it *has* collided is $dt/\tau$, in
which case its post-collision velocity has zero mean. Putting both
contributions together,
$$
\langle\vec v(t+dt)\rangle = \left(1 - \frac{dt}{\tau}\right)
\!\left[\vec v(t) + \frac{\vec F}{m}\,dt\right]
+ \frac{dt}{\tau}\cdot 0 ,
$$
and keeping terms to $O(dt)$ gives the **Drude equation of motion**,
$$
m\,\frac{d\vec v}{dt} = \vec F - \frac{m\vec v}{\tau} .
\tag{1}
$$

This is a Langevin equation with the random kick averaged out: the linear
friction term $-m\vec v/\tau$ encodes the *relaxation* induced by random
collisions. The computation notebook restores the stochastic kick directly
and recovers (1) on the average, confirming this picture.

Every result in the rest of this notebook is a consequence of (1), with
either an electric field, a magnetic field, or both, plugged in for $\vec
F$.

# DC Conductivity and Ohm's Law

We now apply the Drude equation of motion (1) to the simplest case: a
uniform, static **electric field** $\vec E$. Physically, the field tries to
accelerate the electrons forever, but collisions keep resetting them. After
a few relaxation times the two effects balance and the average acceleration
vanishes, $d\vec v/dt = 0$. Setting the left-hand side of (1) to zero with
$\vec F = -e\vec E$ gives the **drift velocity**,
$$
\vec v_d = -\frac{e\tau}{m}\,\vec E ,
\tag{2}
$$
where $-e$ is the electron charge (with $e > 0$). The minus sign just says
that electrons drift opposite to $\vec E$.

The corresponding **current density** is $\vec J = -ne\vec v_d$, where $n$
is the conduction-electron density. Substituting (2),
$$
\boxed{\;\vec J = \sigma_0\,\vec E ,
\qquad \sigma_0 = \frac{n e^2 \tau}{m}\;}
\tag{3}
$$

This is **Ohm's law**, and $\sigma_0$ is the Drude **DC conductivity**.
Notice that we did not assume Ohm's law --- it dropped out of the dynamics
(1). The physical content is that conduction is set by a competition
between *driving* (the field $\vec E$) and *scattering* (the time scale
$\tau$).

A few common rewrites you will see in textbooks:

- The **resistivity** is $\rho = 1/\sigma_0 = m/(ne^2\tau)$. Stronger
  scattering (smaller $\tau$) means higher resistance.
- The **mobility** is $\mu = e\tau/m$, so $\vec v_d = -\mu \vec E$.
- $\mu$ is what you measure in a Hall bar (see (10)), which then lets you
  back out $\tau$ from the room-temperature resistivity.

Equation (3) is the *zero-frequency* limit of a more general response. We
turn next to what happens when $\vec E$ oscillates.

# AC Response

Take a harmonic field $\vec E(t) = \vec E_0\,e^{-i\omega t}$ and look for a
steady-state velocity of the same form, $\vec v(t) = \vec v_0\,e^{-i\omega t}$.
Substituting into (1) with $\vec F = -e\vec E$,
$$
-i\omega m\,\vec v_0 = -e\vec E_0 - \frac{m}{\tau}\,\vec v_0
\quad\Longrightarrow\quad
\vec v_0 = -\frac{e\tau/m}{1-i\omega\tau}\,\vec E_0 .
$$
The corresponding current $\vec J = -ne\vec v_0\,e^{-i\omega t}$ defines the
frequency-dependent **complex conductivity**,
$$
\boxed{\;\sigma(\omega) = \frac{\sigma_0}{1 - i\omega\tau}
= \frac{\sigma_0}{1+\omega^2\tau^2}
+ i\,\frac{\sigma_0\,\omega\tau}{1+\omega^2\tau^2} \;}
\tag{4}
$$
with the same $\sigma_0 = n e^2 \tau/m$ from (3). The real and imaginary
parts have distinct physical meanings:

- $\operatorname{Re}\sigma(\omega)$ controls **dissipative** transport ---
  energy lost to the lattice, i.e. Joule heating (see (16)).
- $\operatorname{Im}\sigma(\omega)$ controls the **reactive** response ---
  energy that sloshes between the field and the electron gas without
  heating anything.

Two limits are worth memorizing:

- *Low frequency*, $\omega\tau \ll 1$: $\sigma(\omega) \to \sigma_0$ ---
  the DC result (3).
- *High frequency*, $\omega\tau \gg 1$: $\sigma(\omega) \to i\,n e^2/(m\omega)$
  --- a purely imaginary, **collisionless** electron-gas response. The drift
  never has time to relax before the field reverses.

The *phase lag* between $\vec J$ and $\vec E$ is $\varphi(\omega) =
\arctan(\omega\tau)$, growing from $0$ at DC to $\pi/2$ in the collisionless
limit.

## Dielectric function and the plasma frequency

The relative permittivity of the electron gas follows from
$\epsilon(\omega) = 1 + i\sigma(\omega)/(\epsilon_0\omega)$. Using (4),
$$
\epsilon(\omega) = 1 - \frac{\omega_p^2}{\omega^2 + i\omega/\tau} ,
\tag{5}
$$
where the **plasma frequency** is defined by
$$
\omega_p^2 \equiv \frac{n e^2}{\epsilon_0 m} .
\tag{6}
$$

In the collisionless limit $\tau \to \infty$, $\epsilon(\omega) = 1 -
\omega_p^2/\omega^2$. The metal is opaque for $\omega < \omega_p$
(reflective: $\epsilon < 0$, evanescent waves) and transparent for
$\omega > \omega_p$ ($\epsilon > 0$, propagating waves). This is why alkali
metals --- whose plasma edges land in the UV --- become UV-transparent
above $\omega_p$.

A useful sanity-check identity: differentiating (4) shows that the
*spectral weight* $\int_0^\infty \operatorname{Re}\sigma(\omega)\,d\omega
= \pi\sigma_0/(2\tau) = \pi n e^2/(2m)$ is independent of $\tau$. This is
the **f-sum rule** of the free electron gas; we recover it numerically in
the computation notebook via Green-Kubo.

# The Hall Effect and Magnetoresistance

Now add a static magnetic field $\vec B = B\hat z$. The Drude equation of
motion picks up the Lorentz force,
$$
m\,\frac{d\vec v}{dt} = -e(\vec E + \vec v\times\vec B) - \frac{m\vec v}{\tau} .
\tag{7}
$$
The electric field drives electrons along the conductor; the magnetic field
deflects them perpendicular to their motion. Charge piles up on the sides
of the sample until a transverse electric field cancels the Lorentz force
--- a steady state --- and that transverse field is the **Hall voltage**.

For steady transport, $\vec v = (v_x, v_y, 0)$ and $d\vec v/dt = 0$.
Solving (7) component-wise and converting to current density via $\vec J =
-ne\vec v$ gives the **resistivity matrix**,
$$
\begin{pmatrix} E_x \\ E_y \end{pmatrix}
= \frac{1}{\sigma_0}
\begin{pmatrix} 1 & \omega_c\tau \\ -\omega_c\tau & 1 \end{pmatrix}
\begin{pmatrix} J_x \\ J_y \end{pmatrix} ,
\tag{8}
$$
where the **cyclotron frequency**
$$
\omega_c \equiv \frac{eB}{m}
\tag{9}
$$
sets the rate at which electrons orbit the magnetic-field direction.

In a Hall-bar geometry one imposes $J_y = 0$ (no transverse current can
escape) and reads off the transverse electric field $E_y$. Equation (8) then
gives the **Hall coefficient**,
$$
\boxed{\;R_H \equiv \frac{E_y}{J_x B} = -\frac{1}{n e}\;}
\tag{10}
$$

Three remarks worth making:

- **$R_H$ measures carrier density.** Knowing $R_H$ gives $n$ directly,
  with no other material parameters needed. Combined with (3), it lets you
  extract $\tau$ from a measured $\sigma_0$.
- **The diagonal entry of (8) is independent of $B$.** So the Drude model
  predicts *zero* magnetoresistance --- the longitudinal resistivity does
  not change in a magnetic field. Real metals show a small but nonzero
  magnetoresistance, a hint at physics beyond Drude.
- **The sign is always negative** in (10), because we chose negative
  carriers. Yet some metals (Be, Zn, Al) exhibit *positive* Hall
  coefficients --- a clue that real metals also carry charge in the form of
  hole-like quasiparticles, which require band theory to describe.

The computation notebook verifies (10) numerically by simulating (7) and
inverting the conductivity tensor.

# Thermal Conductivity and the Wiedemann-Franz Law

So far we have only considered charge transport. The Drude electron gas
also carries heat, and the same scattering time $\tau$ controls both. Treat
the electrons as a classical ideal gas with mean energy $\tfrac{3}{2}k_B T$
per particle and mean thermal speed $\bar v$. Elementary kinetic theory
then gives the **thermal conductivity**,
$$
\kappa = \tfrac{1}{3}\,c_v\,\bar v\,\ell ,
\tag{11}
$$
where $c_v = \tfrac{3}{2}n k_B$ is the classical electronic specific heat
per unit volume and $\ell = \bar v\tau$ is the mean free path. Substituting
$\tfrac{1}{2}m\bar v^2 = \tfrac{3}{2}k_B T$ collapses (11) to
$\kappa = \tfrac{3}{2}\,n k_B^2 T \tau / m$.

Dividing by $\sigma_0 T$ (with $\sigma_0$ from (3)) eliminates the unknown
$\tau$ entirely and gives the **Wiedemann-Franz ratio**,
$$
\boxed{\;\frac{\kappa}{\sigma_0 T} \equiv L =
\frac{3}{2}\!\left(\frac{k_B}{e}\right)^{\!2} \;}
\tag{12}
$$

The remarkable feature of (12) is that the right-hand side is a
*universal constant* --- it depends only on $k_B$ and $e$, not on $n$ or
$\tau$ or anything else about the metal. Drude found numerical agreement
with copper and silver to within a factor of two, which he famously
misreported as *exact* agreement after making two algebra errors that
canceled. We will see in (21) that Sommerfeld's correction multiplies the
right-hand side by $\pi^2/3$, giving $L_0 \approx
2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$ --- the value that is
actually observed.

# Joule Heating --- DC

Consider a conductor of cross-section $A$ and length $L$ carrying a uniform
current $I$ under bias $V = EL$. The mechanical work done by the field on
the electrons, *per unit volume per unit time*, is the dot product of
current density and field,
$$
p = \vec J\cdot\vec E .
\tag{13}
$$
Where does this energy go? In steady state the drift kinetic energy per
electron, $\tfrac{1}{2}m v_d^2$, is constant in time. So all of the work
done by the field is transferred to the lattice during collisions ---
that is what "thermalization at collisions" (assumption 4) means. The work
$e\vec E\cdot\vec v\,dt$ delivered to an electron between collisions is
dumped into the lattice at the *next* collision when the electron's
velocity is reset to a thermal value.

Multiplying (13) by the volume $AL$ recovers the engineering form,
$$
P_{\text{DC}} = JE \cdot AL = (JA)(EL) = IV = I^2 R ,
\tag{14}
$$
with $R = L/(\sigma_0 A)$ the conductor's resistance.

An important conceptual takeaway: **Joule heating is a consequence of
momentum relaxation**. An idealized ballistic conductor with $\tau \to
\infty$ has no scattering, no relaxation, and would *store* energy in the
electron gas rather than dissipating it. The real, finite $\tau$ is what
makes a resistor get hot.

# Joule Heating --- AC

For an oscillating field $\vec E(t) = \vec E_0\cos\omega t$, the
instantaneous dissipated power density is again $p(t) = \vec J(t)\cdot
\vec E(t)$. Writing $\vec J(t) = \operatorname{Re}[\sigma(\omega)\vec E_0\,
e^{-i\omega t}]$ with $\sigma(\omega)$ from (4) and time-averaging over one
period,
$$
\boxed{\;\langle p\rangle = \tfrac{1}{2}\,
\operatorname{Re}\sigma(\omega)\,|\vec E_0|^2
= \frac{1}{2}\,\frac{\sigma_0}{1+\omega^2\tau^2}\,|\vec E_0|^2 \;}
\tag{15}
$$

Two things to notice in (15).

- **Only $\operatorname{Re}\sigma$ dissipates.** The imaginary part of (4)
  --- the reactive piece --- corresponds to energy that flows back and
  forth between the field and the electron gas without ever heating the
  lattice.
- **Resistance falls off at high frequency.** For $\omega\tau \gg 1$ the
  dissipation drops as $1/\omega^2$. Physically, the drift never has time
  to relax before the field reverses; the electrons just slosh in place.
  For RF circuits this is welcome (less power deposited in conductors); in
  microwave cavities it competes with skin-effect enhancement.

The computation notebook reproduces (15) by directly time-averaging
$\vec J\cdot\vec E$ over many periods and confirms that the dissipation
matches $\tfrac{1}{2}\operatorname{Re}\sigma(\omega)|\vec E_0|^2$
across two decades in $\omega\tau$.

# Failures of the Classical Drude Model

The Drude picture reproduces the *form* of many metallic response laws ---
Ohm's law (3), the AC conductivity (4), the Hall coefficient (10), and
Wiedemann-Franz (12). But it gets several *magnitudes* wrong, and a careful
experimentalist cannot ignore them.

- **Specific heat.** Equipartition assigns each electron an internal energy
  $\tfrac{3}{2}k_B T$, giving a molar electronic heat capacity
  $\tfrac{3}{2}R \approx 12.5\,\mathrm{J/(mol\,K)}$ --- comparable to the
  lattice and easily measurable. Experiment on copper gives
  $c_v^{(e)} = \gamma T$ with
  $\gamma \approx 0.695\,\mathrm{mJ/(mol\,K^2)}$: two orders of magnitude
  smaller, and *linear* in $T$, not constant.

- **Wiedemann-Franz numerical coefficient.** The Drude Lorenz number from
  (12), $L_\text{Drude} = \tfrac{3}{2}(k_B/e)^2 \approx
  1.11\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$, is off from the
  experimental value
  $L_\text{exp} \approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}}$
  by a factor of $\pi^2/4.5 \approx 2.2$.

- **Temperature dependence of $\rho$.** A geometric ion-cross-section
  argument gives $\tau \propto 1/\sqrt{T}$ and hence
  $\rho \propto \sqrt{T}$. Measurements give $\rho \propto T$ at high $T$
  and $\rho \propto T^5$ at low $T$ (Bloch-Gr\"uneisen).

- **Sign of the Hall coefficient.** Several metals (Be, Zn, Al) show
  $R_H > 0$, impossible for negatively charged free carriers. Band theory
  resolves this through hole-like carriers.

- **Mean electron speed.** Equipartition gives
  $\bar v \sim 10^5\,\mathrm{m/s}$; Hall-bar measurements infer effective
  speeds $\sim 10^6\,\mathrm{m/s}$, *independent of temperature* --- a
  clear sign the relevant speed is a quantum zero-point quantity, not a
  thermal average.

- **Thermopower.** The classical Seebeck coefficient
  $S \sim -k_B/e \approx -86\,\mu\mathrm{V/K}$ is two orders of magnitude
  larger than the measured $1$--$10\,\mu\mathrm{V/K}$.

Each of these discrepancies is repaired by a single change of ingredient:
replace Maxwell-Boltzmann with **Fermi-Dirac** statistics. That is the next
section.

# Fermi-Dirac Statistics and the Sommerfeld Correction

Electrons are spin-$\tfrac{1}{2}$ **fermions**, so the **Pauli exclusion
principle** forbids two of them from occupying the same quantum state
$(\vec k, s)$. At finite temperature the average occupation of a state with
energy $\varepsilon$ is the **Fermi-Dirac distribution**,
$$
f(\varepsilon) = \frac{1}{e^{(\varepsilon - \mu)/k_B T} + 1} ,
\tag{16}
$$
where $\mu$ is the chemical potential.

At $T = 0$, $f$ is a sharp step function: every state up to the **Fermi
energy** $\varepsilon_F$ is filled, every state above it is empty. The
relevant Fermi-surface quantities are
$$
k_F = (3\pi^2 n)^{1/3} ,\qquad
v_F = \frac{\hbar k_F}{m} ,\qquad
T_F = \frac{\varepsilon_F}{k_B} .
\tag{17}
$$
For copper ($n = 8.5\times 10^{28}\,\mathrm{m^{-3}}$),
$T_F \approx 81{,}000\,\mathrm{K}$. At room temperature the dimensionless
temperature
$$
\theta \equiv \frac{T}{T_F} = \frac{k_B T}{\varepsilon_F} \approx 0.004
$$
is *tiny*. We say the electron gas is **strongly degenerate**: thermal
excitations only stir up a thin shell of width $\sim k_B T$ around the
Fermi surface, and only those electrons participate in transport.

## The Sommerfeld expansion

For any smooth function $H(\varepsilon)$, integration by parts of (16)
yields
$$
\int_0^\infty H(\varepsilon)\,f(\varepsilon)\,d\varepsilon =
\int_0^\mu H(\varepsilon)\,d\varepsilon
+ \frac{\pi^2}{6}(k_B T)^2\,H'(\mu) + O(T^4) .
\tag{18}
$$
The correction is suppressed by $(k_B T/\varepsilon_F)^2 \sim 10^{-5}$ for
copper at room temperature --- which is why the *form* of every Drude
formula survives untouched.

## Electronic specific heat

Apply (18) to $H = \varepsilon\,g(\varepsilon)$ where
$g(\varepsilon) = (1/2\pi^2)(2m/\hbar^2)^{3/2}\sqrt{\varepsilon}$ is the
free-electron density of states. The mean energy density is
$u \approx u_0 + \tfrac{\pi^2}{6}(k_B T)^2 g(\varepsilon_F)$, with
$u_0 = \tfrac{3}{5}n\varepsilon_F$ the ground-state energy. Differentiating
in $T$ gives the **electronic specific heat per unit volume**,
$$
\boxed{\;c_v = \frac{\pi^2}{3}\,n k_B\,\frac{k_B T}{\varepsilon_F}
= \frac{\pi^2}{2}\,n k_B\,\theta \;}
\tag{19}
$$
which is *linear* in $T$ and smaller than the classical Drude value by a
factor $\tfrac{\pi^2}{3}\theta \approx 0.033$ at room temperature ---
matching the measured $\gamma T$ behaviour to within a few percent.

## Corrected Lorenz number

Replace $\bar v \to v_F$ and $c_v \to \tfrac{\pi^2}{3}n k_B (k_B T/
\varepsilon_F)$ in the kinetic-theory estimate (11). The unknown $\tau$
again cancels when forming $\kappa/(\sigma_0 T)$, leaving
$$
\boxed{\;\frac{\kappa}{\sigma_0 T} \equiv L_0 =
\frac{\pi^2}{3}\!\left(\frac{k_B}{e}\right)^{\!2}
\approx 2.44\times 10^{-8}\,\mathrm{W\,\Omega\,K^{-2}} \;}
\tag{20}
$$
in agreement with copper, silver, and gold to within a few percent. The
factor $\pi^2/3$ on the right of (20) versus $3/2$ on the right of (12) is
*exactly* the ratio $L_0/L_\text{Drude} \approx 2.20$ flagged in the
failures section.

## Why the conductivity formula is unchanged

A field $\vec E$ shifts the *entire* Fermi sphere by $\delta\vec k =
-e\vec E\tau/\hbar$. The resulting current density,
$\vec J = -(e/V)\sum_{\vec k}\hbar\vec k\,f(\varepsilon_{\vec k})$,
depends only on $n$ and $\tau$ --- not on the *shape* of the occupation
$f$. Hence
$$
\sigma(\omega) = \frac{\sigma_0}{1 - i\omega\tau} ,\qquad
\sigma_0 = \frac{n e^2 \tau}{m} ,
\tag{21}
$$
identical in form to (3) and (4). The Drude formula was *accidentally
right*: correct form, wrong physical reason. Only the *interpretation* of
$\tau$ changes --- it is now the relaxation time of electrons near the
Fermi surface, where the actual excitations live.

The remaining failures (thermopower, $\rho(T)$, Hall sign) require going
beyond free electrons entirely --- band structure and phonons --- which is
the subject of the final section.

# The Drude-Sommerfeld Model

We can now state the **Drude-Sommerfeld** model in one sentence: keep the
Drude equation of motion (1) exactly as it is, and replace every
Maxwell-Boltzmann ensemble average by a Fermi-Dirac one (16). Equation (18)
makes that replacement systematic, and equations (19)-(21) summarize the
quantitative consequences.

What this section adds is *interpretation* --- the physical picture of why
the same dynamics, with a different statistical weight, gives the right
magnitudes.

## What changes, and what stays the same

The conductivity (21) is identical in form to the Drude (3)-(4). The
Wiedemann-Franz law (20) acquires the factor $\pi^2/3$ in place of $3/2$
in (12). The specific heat (19) becomes linear in $T$ instead of constant.
The Hall coefficient (10) is unchanged in form. Every formula in which
$\tau$ appears is now interpreted as a property of the *Fermi-surface*
electrons, not the bulk gas.

## The mechanism of dissipation, revisited

In the Drude picture, Joule heating (15) was said to come from each
electron handing its drift energy to the lattice at the next collision. In
the Drude-Sommerfeld picture the *only* electrons that can be scattered
are those within $\sim k_B T$ of the Fermi surface, because all the states
deeper inside are already full --- the Pauli principle forbids them from
landing somewhere occupied. So a small minority of electrons carries the
entire current and the entire dissipation.

The macroscopic power balance (15) is unchanged --- which is exactly why
Drude could not have detected the difference from a single resistor
measurement. He needed thermodynamics ($c_v$, the Seebeck effect) to see
the failure.

## A useful order of magnitude

In the degenerate limit $\theta \ll 1$, only a *fraction*
$\sim \theta = T/T_F \approx 0.004$ of the electrons are thermally
"active". This single number explains essentially every numerical factor
that the Sommerfeld correction restores: the specific heat (19) is
suppressed by $\theta$, the thermal current is carried only by those
active electrons, and the chemical potential equals $\varepsilon_F$ to
within a correction of order $\theta^2$.

# Beyond Drude-Sommerfeld

The Drude-Sommerfeld picture inherits a few simplifying lies from its
parent that must eventually be paid for. Three are worth naming explicitly:

- **The relaxation time $\tau$ is assumed independent of $\vec k$.** In real
  metals $\tau(\vec k)$ varies over the Fermi surface, and different
  scattering channels (phonons, impurities, electron-electron) add via
  **Matthiessen's rule**. This is what gives $\rho(T)$ its correct
  temperature dependence.
- **The band structure is assumed parabolic.** Replacing $m$ by an
  **effective mass** $m^*$ absorbs much but not all of the discrepancy. Open
  orbits and multiple bands produce qualitatively new effects (magnetic
  breakdown, the *positive* Hall coefficients of Be, Zn, and Al).
- **Electron-electron interactions are ignored.** Landau's **Fermi-liquid
  theory** shows that at low temperatures these interactions can be
  absorbed into renormalized parameters ($m^*$, $v_F^*$, $\tau^*$), leaving
  the Drude-Sommerfeld *form* intact down to remarkably low temperatures.

The corresponding hierarchy of models is
$$
\text{Drude} \;\to\; \text{Drude-Sommerfeld} \;\to\;
\text{Bloch-Boltzmann} \;\to\; \text{Fermi liquid} ,
$$
each step adding one ingredient: quantum statistics (16), then a periodic
potential, then quasiparticle interactions. Our computation notebook
stops at the second step --- the Drude-Sommerfeld model --- and verifies
it numerically.

| Mean speed | $\sqrt{3k_B T/m}$ | $v_F$ | $\sim 10^6\,\mathrm{m/s}$ |
| $R_H$ (Eq. 10) | $-1/(n e)$ | $-1/(n e)$ | sign wrong for Be, Zn, Al |

The Drude model is right about the *structure* of metallic response ---
linear Ohmic conduction, a Lorentzian conductivity edge at $\omega\tau
\sim 1$, Joule heating proportional to $|E|^2$ --- and wrong about any
quantity that depends on *counting* how many electrons actually move, or
with what speed. The Sommerfeld correction is, in a precise sense, the
minimal repair: replace classical counting by quantum counting, and leave
the dynamics alone. The resulting model is accurate enough that it remains
the standard starting point for metal and plasma electrodynamics well into
the twenty-first century.

from the Fermi-Dirac distribution (16) --- recovering both the corrected
specific heat (19) and the experimental Lorenz number (20).
